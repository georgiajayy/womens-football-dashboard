```python
matches = sb.matches(competition_id=53, season_id=315)
matches.shape
```

```python
england = matches[
    matches["home_team"].str.contains("England")
    matches["away_team"].str.contains("England")
    ]
enfland[["match_id", "match_date", "home_team", "home_score", "away_score", "away_team"]].sort_values("match_date")
```

```python
england = matches[
    matches["home_team"].str.contains("England") |
    matches["away_team"].str.contains("England")
]
england[["match_id", "match_date", "home_team", "home_score", "away_score", "away_team"]].sort_values("match_date")
```
